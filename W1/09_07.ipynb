```python
def computeIoU(boxA, boxB):
    xA = max(boxA[0], boxB[0])
    yA = max(boxA[1], boxB[1])
    xB = min(boxA[2], boxB[2])
    yB = min(boxA[3], boxB[3])
    interArea = max(0, xB - xA + 1) * max(0, yB - yA + 1)
    boxAArea = (boxA[2] - boxA[0] + 1) * (boxA[3] - boxA[1] + 1)
    boxBArea = (boxB[2] - boxB[0] + 1) * (boxB[3] - boxB[1] + 1)
    iou = interArea / float(boxAArea + boxBArea - interArea)
    return iou
boxA = [0, 0, 100, 100]
boxB = [50, 50, 150, 150]
print(computeIoU(boxA, boxB))
```

```python
#SIDM: single instruction, multiple data
def non_maximum_suppression(boxes, scores, iou_threshold):
    sorted_indices = sorted(range(len(scores)), key = lambda x : -scores[x])
    kept_indices = []
    while sorted_indices:
        index = sorted_indices.pop(0)
        kept_indices.append(index)
        selected_indices = []
        for i in sorted_indices:
            iou = computeIoU(boxes[index], boxes[i])
            if iou <= iou_threshold:
                selected_indices.append(i)
        sorted_indices = selected_indices
    return kept_indices

boxes = [
    [100, 100, 210, 210],
    [105, 105, 215, 215],
    [150, 150, 250, 250]
]
scores = [0.9, 0.8, 0.7]
print(non_maximum_suppression(boxes, scores, 0.7))
```

```python
def non_maximum_suppression_array(boxes, scores, threshold):
    if len(boxes) == 0:
        return []
    x1 = boxes[:,0]
    y1 = boxes[:,1]
    x2 = boxes[:,2]
    y2 = boxes[:,3]
    areas = (x2 - x1 + 1) * (y2 - y1 + 1)
    keep = []
    order = scores.argsort()[::-1]
    while order.size > 0:
        i = order[0]
        keep.append(i)
        xx1 = np.maximum(x1[i], x1[order[1:]])
        yy1 = np.maximum(y1[i], y1[order[1:]])
        xx2 = np.minimum(x2[i], x2[order[1:]])
        yy2 = np.minimum(y2[i], y2[order[1:]])
        w = np.maximum(0, xx2 - xx1 + 1)
        h = np.maximum(0, yy2 - yy1 + 1)
        intersection = w * h
        iou = intersection / (areas[i] + areas[order[1:]] - intersection)
        inds = np.where(iou <= threshold)[0]
        print(inds)
        order = order[inds + 1]
    return keep
boxes = np.array(boxes)
scores = np.array(scores)
threshold = 0.3
print(boxes[non_maximum_suppression_array(boxes, scores, threshold)])
```
