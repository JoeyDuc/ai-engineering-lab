```python
#repeat(data, repeats, axis)

data = np.array([1, 4])
data = np.repeat(data, 2)
print(data)
print("\n\n")


data = np.arange(100).reshape(10, 10)
print(data)
print('\n\n')
#vertical: 0, horizontal: 1

data = np.repeat(data, 2, axis = 0)
print(data)
print('\n\n')
data = np.repeat(data, 2, axis = 1)
print(data)
```

```python
#Aray Indexing

#Slicing
data = np.array([
    [1, 2],
    [3, 4],
    [5, 6]
])
print(data, end = "\n\n")
print(data[1:3], end = "\n\n")
print(data[0:1, 0], end = "\n\n")
print(data[:], end = "\n\n")
```

```python
data = np.arange(4).reshape(2, 2)
print(data, end = '\n\n')

print(np.sum(data, axis = 0), end = '\n\n')

print(np.sum(data, axis = 1), end = '\n\n')

print(np.mean(data), end = '\n\n')

print(np.mean(data, axis = 0), end = '\n\n')

print(np.mean(data, axis = 1), end = '\n\n')

print(np.max(data, axis = 0), end = '\n\n')

print(np.max(data, axis = 1), end = '\n\n')
```

```python
#Broadcasting
end = '\n\n'
data = np.arange(1, 4)
print(data, end = end)


print(data + 10, end = end)

print(data - 10, end = end)
```

```python
data = np.arange(1, 10).reshape(3, 3)
print(data, end = end)

b = np.array([1, 2, 3])
print(data + b, end = end)
```

```python
data = np.array([1, 2, 3])

print(data, end = end)

max_value = data.max()
print(max_value, end = end)

data_exp = np.exp(data)
print(data_exp, end = end)

sum_data_exp = np.sum(data_exp)
print(sum_data_exp, end = end)

result = data_exp / sum_data_exp
print(result, end = end)
```
