```python
import matplotlib.pyplot as plt
import cv2
image = cv2.imread('oggy.jpg', 1)

print(image.dtype)

#process
image = image.astype(float)
image = image + 50
image = image[:, :, ::-1]

image = np.clip(image, a_min = 0, a_max = 255)
image = image.astype(np.uint8)
print(image.dtype)
plt.imshow(image)
```

```python
data = np.array([
    [2, 6, 0, 1],
    [3, 0, 8, 3],
    [4, 4, 5, 0],
    [2, 0, 4, 6]
])
print('data:\n', data)

result0 = np.apply_along_axis(np.max, axis = 0, arr = data)
print('result0: \n', result0)

result1 = np.apply_along_axis(np.median, axis = 1, arr = data)
print('result1: \n', result1)

result2 = np.apply_along_axis(np.max, arr = data, axis = 0)
print('result2: \n', result2)
```

```python
def compute_average(vector):
    return vector.mean()
img = cv2.imread('7.jpg', 1)
img = np.apply_along_axis(compute_average, axis = 2, arr = img)
print(img.shape)
plt.imshow(img, cmap='gray')
```

```python
def compute_sum(vector):
    return vector.sum()
data = np.ones((3, 4, 5))
print(data)
result = np.apply_along_axis(compute_sum, axis=1, arr = data)
print(result.shape)
print(result)
```
