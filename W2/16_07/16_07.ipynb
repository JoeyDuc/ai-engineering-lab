```python
#Image Similarity
import numpy as np
from PIL import Image
image1 = Image.open('data/image1.jpg')
image2 = Image.open('data/image2.jpg')
query_image = Image.open('data/query_image.jpg')

print("Size:")
print(image1.size)
print(image2.size)
print(query_image.size)

img1_data = np.array(image1)
img2_data = np.array(image2)
query_img_data = np.array(query_image)

print("Shape: ")
print(img1_data.shape)
print(img2_data.shape)
print(query_img_data.shape)
```

```python
def compute_average(vector):
    return vector.mean()
img1_gray = np.apply_along_axis(compute_average, axis = 2, arr = img1_data)
img2_gray = np.apply_along_axis(compute_average, axis = 2, arr = img2_data)
query_img_gray = np.apply_along_axis(compute_average, axis = 2, arr = query_img_data)

img1_flatten = img1_gray.flatten()
img2_flatten = img2_gray.flatten()
query_img_flatten = query_img_gray.flatten()
```

```python
def compute_cosine_similarity(A, B):
    cosine = np.dot(A, B) / ((np.linalg.norm(A)) * (np.linalg.norm(B)))
    return cosine
cosine1 = compute_cosine_similarity(img1_flatten, query_img_flatten)
print("cosine1: ", cosine1)

consine2 = compute_cosine_similarity(img2_flatten, query_img_flatten)
print("cosine2: ", consine2)
```
