```python
background = cv2.imread('3.BackgoundSubtraction/background.png', 1)
stillImage = cv2.imread('3.BackgoundSubtraction/StillImage.png', 1)
new_bg = cv2.imread('3.BackgoundSubtraction/FakeBackground.png')
background = cv2.resize(background, (640, 480))
stillImage = cv2.resize(stillImage, (640, 480))
new_bg = cv2.resize(new_bg, (640, 480))


difference = cv2.absdiff(background, stillImage)
_, difference_binary = cv2.threshold(difference, 15, 255, cv2.THRESH_BINARY)
output = np.where(difference_binary == 0, new_bg, stillImage)
plt.imshow(output[:, :, ::-1])
```

```python
u = np.array([3, 2, 4, 1])
v = np.array([2, 1, 0, 3])

r = np.einsum('i,i->', u, v)
print(r)

r = np.einsum('i, j->ij', u, v)
print(r)
```

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print(np.einsum('i,j->ji', a, b))
print(np.einsum('i,j->i', a, b))
```

```python
A = np.array([
    [5, 1, 2],
    [3, 1, 6],
    [2, 4, 3]
])
print(np.einsum('ii->i', A))

print(np.einsum('ii->', A))
```

```python
A = np.array([
    [3, 0, 1],
    [2, 2, 3]
])
B = np.array([
    [1, 0, 2],
    [3, 2, 1]
])

r = np.einsum('ij,ij->ij', A, B)
print(r)
```

```python
A = np.array([
    [0, 2, 1],
    [3, 1, 4]
])
B = np.array([
    [2, 5],
    [0, 1],
    [3, 4]
])

C = np.einsum('ij, jk -> ik', A, B)
print(C)
```
