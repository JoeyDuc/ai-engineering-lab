```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
data = pd.read_csv('vi_text_retrieval.csv')
print(data)
```

```python
data.iloc[0]
```

```python
data['text']
```

```python
#Convert to Lower
context = data['text']
context = [doc.lower() for doc in context]
context[:5]
```

```python
tf_idf_vectorizer = TfidfVectorizer(max_features=500)
```

```python
context_embeded = tf_idf_vectorizer.fit_transform(context)
```

```python
context_embeded
```

```python
tf_idf_vectorizer.vocabulary_
```

```python
question = data.iloc[0]['question']
print(question)
```

```python
query_embeded = tf_idf_vectorizer.transform([question.lower()])
query_embeded.toarray()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
consine_scores = cosine_similarity(context_embeded, query_embeded).reshape((-1, ))

consine_scores
```

```python
top_d = 5
for idx in consine_scores.argsort()[-top_d:][::-1]:
    print('Score: ',consine_scores[idx])
    print(data.iloc[idx, 2])
    print('______')
```

```python
corr_coef_scores = np.corrcoef(query_embeded.toarray()[0], context_embeded.toarray())
for index in corr_coef_scores[0][1:].argsort()[-top_d:][::-1]:
    print("Scores", corr_coef_scores[0][index])
    print(data.iloc[index, 2])
    print("_______________")
```
