```python
def calculate_median(numbers):
    N = len(numbers)
    numbers.sort()
    if N % 2 == 0:
        m1 = N // 2
        m2 = (N // 2) + 1
        median = (numbers[m1 - 1] + numbers[m2 - 1]) / 2
    else:
        median = numbers[N // 2 - 1]
    return median
```

```python
data = [1, 4, 2, 5, 2]
print(calculate_median(data))

data = [1, 3, 5, 7, 9, 2]
print(calculate_median(data))
```

```python
import numpy as np

x = np.array([5, 3, 6, 7, 4])

var = (x - np.mean(x)) ** 2

var = np.mean(var)
print(var)

standard_deviation = np.sqrt(var)

print(standard_deviation)
```

```python
#variance
def calculate_mean(numbers):
    s = sum(numbers)
    N = len(numbers)
    mean = s / N
    return mean
def calculate_variance(numbers):
    mean = calculate_mean(numbers)

    diff = []
    for num in numbers:
        diff.append(num - mean)

    squared_diff = []
    for d in diff:
        squared_diff.append(d ** 2)

    sum_squared_diff = sum(squared_diff)
    variance = sum_squared_diff / len(numbers)

    return variance


numbers = [5, 3, 6, 7, 4]
print(calculate_mean(numbers))

print(calculate_variance(numbers))

print('standard deviation = :', calculate_variance(numbers) ** 0.5)
    

    
```
